```python
import pandas as pd
df = pd.read_csv('식당 평점 및 리뷰 수.csv')
df
```

```python
df_dobong = pd.read_csv('도봉구_혼밥식당.csv')
df_dobong['시도명'] = df_dobong['지역명'].apply(lambda x: x.split(' ')[0])
df_dobong['시군구'] = df_dobong['지역명'].apply(lambda x: x.split(' ')[1])

# 정보없음 떨구기
df_dobong = df_dobong[(df_dobong['평점'] != '정보 없음') & (df['방문자 리뷰 수'] != '정보 없음') & (df_dobong['블로그 리뷰 수'] != '정보 없음')].reset_index(drop=True)

# 문자형 -> 수치형 변수 전환
df_dobong['평점'] = df_dobong['평점'].apply(lambda x: x.replace('별점\n',''))
df_dobong['방문자 리뷰 수'] = df_dobong['방문자 리뷰 수'].apply(lambda x: x.replace('방문자 리뷰',''))
df_dobong['방문자 리뷰 수'] = df_dobong['방문자 리뷰 수'].apply(lambda x: x.replace(',',''))

df_dobong['블로그 리뷰 수'] = df_dobong['블로그 리뷰 수'].apply(lambda x: x.replace('블로그 리뷰',''))
df_dobong['블로그 리뷰 수'] = df_dobong['블로그 리뷰 수'].apply(lambda x: x.replace(',',''))

df_dobong['평점'] = df_dobong['평점'].astype('float64') 
df_dobong['방문자 리뷰 수'] = df_dobong['방문자 리뷰 수'].astype('int64') 
df_dobong['블로그 리뷰 수'] = df_dobong['블로그 리뷰 수'].astype('int64')

df_dobong
```

```python
feat_df = df_dobong.groupby(['시도명','시군구'])[['평점','방문자 리뷰 수','블로그 리뷰 수']].mean().reset_index()
feat_df
```

```python
df = pd.concat([df,feat_df])
```

```python
import pandas as pd
df_yj = pd.read_csv('연제구_혼밥식당.csv')

df_yj['시도명'] = df_yj['지역명'].apply(lambda x: x.split(' ')[0])
df_yj['시군구'] = df_yj['지역명'].apply(lambda x: x.split(' ')[1])

# 정보없음 떨구기
df_yj = df_yj[(df_yj['평점'] != '정보 없음') & (df_yj['방문자 리뷰 수'] != '정보 없음') & (df_yj['블로그 리뷰 수'] != '정보 없음')].reset_index(drop=True)

# 문자형 -> 수치형 변수 전환
df_yj['평점'] = df_yj['평점'].apply(lambda x: x.replace('별점\n',''))
df_yj['방문자 리뷰 수'] = df_yj['방문자 리뷰 수'].apply(lambda x: x.replace('방문자 리뷰',''))
df_yj['방문자 리뷰 수'] = df_yj['방문자 리뷰 수'].apply(lambda x: x.replace(',',''))

df_yj['블로그 리뷰 수'] = df_yj['블로그 리뷰 수'].apply(lambda x: x.replace('블로그 리뷰',''))
df_yj['블로그 리뷰 수'] = df_yj['블로그 리뷰 수'].apply(lambda x: x.replace(',',''))

df_yj['평점'] = df_yj['평점'].astype('float64') 
df_yj['방문자 리뷰 수'] = df_yj['방문자 리뷰 수'].astype('int64') 
df_yj['블로그 리뷰 수'] = df_yj['블로그 리뷰 수'].astype('int64')

df_yj
```

```python
gb_df = df_yj.groupby(['시도명','시군구'])[['평점','방문자 리뷰 수','블로그 리뷰 수']].mean().reset_index()
gb_df
```

```python
df = pd.concat([df,gb_df])
```
