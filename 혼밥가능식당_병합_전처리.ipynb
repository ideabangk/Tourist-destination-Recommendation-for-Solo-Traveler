```python
import pandas as pd
df1 = pd.read_csv('혼밥식당_크롤링_0_to_737.csv')
df2 = pd.read_csv('혼밥식당_크롤링_737_to_1473.csv')
df3 = pd.read_csv('혼밥식당_크롤링_1474_to_2210.csv')
df4 = pd.read_excel('혼밥식당_크롤링_2210_to_end.xlsx')
df_dobong =
```

```python
df2 = df2.loc[:,'지역명':]
df3 = df3.loc[:,'지역명':]
df4 = df4.loc[:,'지역명':]
```

```python
df = pd.concat([df1, df2,df3,df4]).reset_index(drop=True)
```

```python
# 정보없음 떨구기
df = df[(df['평점'] != '정보 없음') & (df['방문자 리뷰 수'] != '정보 없음') & (df['블로그 리뷰 수'] != '정보 없음')].reset_index(drop=True)
# 세종시 떨구기
df = df[~df['지역명'].str.contains('세종특별자치시')].reset_index(drop=True) 
```

```python
# 문자형 -> 수치형 변수 전환
df['평점'] = df['평점'].apply(lambda x: x.replace('별점\n',''))
df['방문자 리뷰 수'] = df['방문자 리뷰 수'].apply(lambda x: x.replace('방문자 리뷰',''))
df['방문자 리뷰 수'] = df['방문자 리뷰 수'].apply(lambda x: x.replace(',',''))

df['블로그 리뷰 수'] = df['블로그 리뷰 수'].apply(lambda x: x.replace('블로그 리뷰',''))
df['블로그 리뷰 수'] = df['블로그 리뷰 수'].apply(lambda x: x.replace(',',''))

df['평점'] = df['평점'].astype('float64') 
df['방문자 리뷰 수'] = df['방문자 리뷰 수'].astype('int64') 
df['블로그 리뷰 수'] = df['블로그 리뷰 수'].astype('int64')

df
```

```python
#df.to_csv('혼밥가능식당_시각화용.csv',index=False)
```

```python
df['시도명'] =  df['지역명'].apply(lambda x: x.split(' ')[0])
df['시군구'] = df['지역명'].apply(lambda x: x.split(' ')[1])
df
```

```python
feat_df = df.groupby(['시도명','시군구'])[['평점','방문자 리뷰 수','블로그 리뷰 수']].mean().reset_index()
feat_df
```
